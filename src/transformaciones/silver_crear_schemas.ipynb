# Schemas de transformación

Crea los schemas de salida solo si no existen. Esta operación es segura para incluirla en cada ejecución del pipeline.

```python
dbutils.widgets.text("catalog", "steam_game_intelligence")
dbutils.widgets.text("silver_schema", "silver")
dbutils.widgets.text("dlq_schema", "dlq")
dbutils.widgets.text("audit_schema", "audit")
catalog = dbutils.widgets.get("catalog")
silver_schema = dbutils.widgets.get("silver_schema")
dlq_schema = dbutils.widgets.get("dlq_schema")
audit_schema = dbutils.widgets.get("audit_schema")
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:catalog || '.' || :silver_schema);
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:catalog || '.' || :dlq_schema);
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:catalog || '.' || :audit_schema);
```
